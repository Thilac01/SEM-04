```python
import numpy as np
from scipy.stats import beta


a,b = 2,8

mean, var = beta.stats(a, b, moments='mv')

print(f"Mean: {mean}, Variance: {var}")

density = beta.pdf(0.8, a, b)
print(f"PDF at 0.8: {density}")
```

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 500)

# Low data (Wide curve)
plt.plot(x, beta.pdf(x, 2, 8), label='a=2, b=8 (Low Data)', lw=2)

# Medium data
plt.plot(x, beta.pdf(x, 20, 80), label='a=20, b=80', lw=2)

# High data (The "Spike")
plt.plot(x, beta.pdf(x, 200, 800), label='a=200, b=800 (High Data)', lw=2)

plt.title('Beta Distribution: How Data Creates a Spike')
plt.xlabel('Probability (x)')
plt.ylabel('Density')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
